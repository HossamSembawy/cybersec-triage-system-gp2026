## Email Dataset — Phishing Email Dataset
**Source:** Kaggle — `naserabdullahalam/phishing-email-dataset`
**Corpora:** Enron, Ling, CEAS 2008, Nazario, Nigerian Fraud, SpamAssassin
**Module:** Email phishing detection (DistilBERT)

Exploring class balance, null values, and text structure of the combined dataset.

```python
import pandas as pd

df = pd.read_csv("/content/sample_data/phishing_email.csv")

print(f"Total rows: {len(df):,}")
print(f"Columns: {list(df.columns)}")
print(f"\nClass balance:")
print(df['label'].value_counts())
print(f"\nClass %:")
print(df['label'].value_counts(normalize=True).mul(100).round(2))
print(f"\nNulls:")
print(df.isnull().sum())
print(f"\nSample text_combined[0]:")
print(df['text_combined'].iloc[0][:300])
```

## URL Dataset — Malicious URLs Dataset
**Source:** Kaggle — `sid321axn/malicious-urls-dataset`
**Origin:** Built on ISCX-URL-2016 (UNB CIC)
**Module:** URL threat classification (Random Forest / XGBoost)

Exploring class distribution across four URL types: benign, phishing, defacement, malware.
Key concerns: multi-class imbalance, lexical feature engineering required from raw URL strings.

```python
import pandas as pd

df = pd.read_csv("/content/sample_data/malicious_phish.csv")

print(f"Total rows: {len(df):,}")
print(f"Columns: {list(df.columns)}")
print(f"\nClass balance:")
print(df['type'].value_counts())
print(f"\nClass %:")
print(df['type'].value_counts(normalize=True).mul(100).round(2))
print(f"\nNulls:")
print(df.isnull().sum())
print(f"\nSample URLs:")
print(df['url'].head(10))
```
